```python
# Melakukan input variable nama dan nilai untuk 5 mahasiswa
for i in range(1, 6):
    print(f"\n--- Mahasiswa-{i} ---")
    nama = input("Masukkan nama mahasiswa: ")
    print("Nama mahasiswa:", nama)

    # Default kan variable total tugas menjadi 0
    total_tugas = 0

    # Melakukan Looping dengan memasukkan nilai tugas 1 dan tugas 2
    for j in range(1, 3):
        tugas = float(input(f"Masukkan nilai tugas {j}: "))
        total_tugas += tugas

    # Setelah dilakukan proses Looping dan penjumlahan dilakukan pembagian dua dari variable tugas agar menjadi total tugas
    tugas_total = total_tugas / 2

    # Dilanjutkan dengan input nilai UTS dan UAS
    uts = float(input("Masukkan nilai UTS: "))
    uas = float(input("Masukkan nilai UAS: "))

    # Menghitung total nilai dengan kontribusi tugas_total (20%), uts (30%), dan uas (50%)
    total = (tugas_total * 0.2) + (uts * 0.3) + (uas * 0.5)

    # Tentukan Grade
    if total >= 80:
        grade = "A"
    elif total >= 70:
        grade = "B"
    elif total >= 60:
        grade = "C"
    elif total >= 50:
        grade = "D"
    else:
        grade = "E"

    # Tampilkan hasil untuk mahasiswa saat ini
    print("\n--- Hasil Penilaian ---")
    print("Nama mahasiswa:", nama)
    print("Nilai Tugas (Rata-rata):", tugas_total)
    print("Nilai UTS:", uts)
    print("Nilai UAS:", uas)
    print("Total nilai:", total)
    print("Nilai Akhir (Grade):", grade)
```
